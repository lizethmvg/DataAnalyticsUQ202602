## Ejercicios
**1. Escriba una consulta sobre la tabla indicadores_paises que:**

- Calcule el PIB Per Cápita estimado en dólares multiplicando pib_billones_usd * 1000000000 y dividiéndolo por poblacion_millones * 1000000. Redondea el resultado a 2 decimales y asígnale el alias pib_per_capita_usd.

- Convierta la columna subsidios_millones a tipo DECIMAL(10,2) y utilice COALESCE(CAST(subsidios_millones AS DECIMAL(10,2)), 0) para mostrar 0.00 cuando no haya registros de subsidios.

- Asigne el alias subsidios_reportados

**2. Comparativo Macroeconómico por Bloque (GROUP BY + HAVING) Consigna:**

- Genera un informe para un análisis de política económica agrupado por region_economica que muestre:

- El total de población sumada por región (SUM(poblacion_millones)).

- La tasa de desempleo más alta registrada dentro de la región (MAX(CAST(tasa_desempleo_pct AS DECIMAL(5,2)))).

- La tasa de desempleo más baja registrada dentro de la región (MIN(CAST(tasa_desempleo_pct AS DECIMAL(5,2)))).
- El promedio de la tasa de inflación regional redondeado a 2 decimales (ROUND(AVG(tasa_inflacion_pct), 2)).

- Muestra únicamente aquellas regiones económicas cuyo PIB acumulado sea mayor o igual a 3.0 billones de dólares.

## DESARROLLO

**CONSULTA 1**

```python
%sql
SELECT
    pais,
    ROUND((pib_billones_usd * 1000000000) / (poblacion_millones * 1000000), 2) AS pib_per_capita_usd,
    COALESCE(CAST(subsidios_millones AS DECIMAL(10,2)), 0) AS subsidios_reportados
FROM workspace.electiva.indicadores_paises;
```

**CONSULTA 2**

```python
%sql
SELECT
    region_economica,
    SUM(poblacion_millones) AS poblacion_total,
    MAX(CAST(tasa_desempleo_pct AS DECIMAL(5,2))) AS desempleo_maximo,
    MIN(CAST(tasa_desempleo_pct AS DECIMAL(5,2))) AS desempleo_minimo,
    ROUND(AVG(tasa_inflacion_pct), 2) AS inflacion_promedio
FROM workspace.electiva.indicadores_paises
GROUP BY region_economica
HAVING SUM(pib_billones_usd) >= 3.0;
```
