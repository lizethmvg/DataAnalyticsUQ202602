**PUNTO 1**

Para ingresar al sitio web de la empresa, a los clientes se les proporciona un nombre de 
usuario el cual es una cadena de texto conformada por el customerid asociado a cada 
usuario, seguido de un guión, las primeras 2 letras de firstname (en mayúsculas) ,otro guión 
las primeras 2 letras del  lastname (en minúsculas), otro guión y termina con la cadena 
genérica “AW” (mayúsculas)  que es la misma para todos los clientes. Esta cadena se 
denomina user_name. 
Desarrollar una consulta SQL para obtener una tabla con el customerid, el fullname y el 
user_name de la persona,

```python
%sql
SELECT
    CustomerID AS customerid,
    FullName AS fullname,
    CONCAT(
        CAST(CustomerID AS STRING), '-',
        UPPER(SUBSTRING(FirstName, 1, 2)), '-',
        LOWER(SUBSTRING(LastName, 1, 2)), '-',
        'AW'
    ) AS user_name
FROM workspace.electiva.customers
ORDER BY customerid;
```

**PUNTO 3**

Durante el mes de agosto de cada año, se aplica un descuento dependiendo de la categoría 
del producto de la siguiente manera: 
- - - - 
- 10% a todas las orders dentro de la category_name= ‘Bikes’ 

- 15% a todas las orders dentro de la category_name=’Clothing’ 

- 8% a todas las orders dentro de la category_name=’Components’ 

- 20% a todas las orders dentro de la category_name=’Accessories’ 


Cree una tabla nueva en el esquema. 
En esta tabla se deben mostrar todas las orders realizadas en el mes de agosto de 2011, y 
para cada una de las orders mostrar los campos :   - 
productname - - - - - - 
modelname 
user_name 
salesorderid 
categoryname 
orderdate 
linetotal 
Las columnas anteriormente mencionadas ya vienen en las tablas, además debe 
crear dos columnas adicionales para la vista : 

●  Una columna indicando qué porcentaje de descuento se aplica a esta 
compra por estar en temporada de mitad de año, esta columna se debe 
llamar descuento_mitad_año y debe contener el descuento correspondiente 
que aplique en cada fila, si no le aplica ningún descuento a la orden por no 
pertenecer a una categoría, el descuento es 0 

● Una columna llamada total_descuentos_aplicados la cual se calcula restando 
de la columna linetotal el descuento , es decir 
linetotal-linetotal*descuento_mitad_año. 

```python
%sql
SELECT
    p.ProductName AS productname,
    p.ModelName AS modelname,
    u.user_name AS user_name,
    o.SalesOrderID AS salesorderid,
    c.CategoryName AS categoryname,
    o.OrderDate AS orderdate,
    o.LineTotal AS linetotal
FROM workspace.electiva.orders AS o
JOIN workspace.electiva.page_user AS u ON o.CustomerID = u.customerid
JOIN workspace.electiva.products AS p ON o.ProductID = p.ProductID
JOIN workspace.electiva.product_sub_categories AS s ON p.SubCategoryID = s.SubCategoryID
JOIN workspace.electiva.product_categories AS c ON s.CategoryID = c.CategoryID
WHERE o.OrderDate BETWEEN '2011-08-01' AND '2011-08-31';
```

```python
%sql
CREATE OR REPLACE TABLE workspace.electiva.ordenes_mitad_anio AS
WITH base AS (
    SELECT
        p.ProductName AS productname,
        p.ModelName AS modelname,
        u.user_name AS user_name,
        o.SalesOrderID AS salesorderid,
        c.CategoryName AS categoryname,
        o.OrderDate AS orderdate,
        o.LineTotal AS linetotal
    FROM workspace.electiva.orders AS o
    JOIN workspace.electiva.page_user AS u ON o.CustomerID = u.customerid
    JOIN workspace.electiva.products AS p ON o.ProductID = p.ProductID
    JOIN workspace.electiva.product_sub_categories AS s ON p.SubCategoryID = s.SubCategoryID
    JOIN workspace.electiva.product_categories AS c ON s.CategoryID = c.CategoryID
    WHERE o.OrderDate BETWEEN '2011-08-01' AND '2011-08-31'
)
SELECT
    productname,
    modelname,
    user_name,
    salesorderid,
    categoryname,
    orderdate,
    linetotal,
    CASE
        WHEN categoryname ILIKE 'Bikes' THEN 0.10
        WHEN categoryname ILIKE 'Clothing' THEN 0.15
        WHEN categoryname ILIKE 'Components' THEN 0.08
        WHEN categoryname ILIKE 'Accessories' THEN 0.20
        ELSE 0
    END AS `descuento_mitad_año`,
    linetotal - linetotal * (
        CASE
            WHEN categoryname ILIKE 'Bikes' THEN 0.10
            WHEN categoryname ILIKE 'Clothing' THEN 0.15
            WHEN categoryname ILIKE 'Components' THEN 0.08
            WHEN categoryname ILIKE 'Accessories' THEN 0.20
            ELSE 0
        END
    ) AS total_descuentos_aplicados
FROM base;
```

```python
%sql
SELECT * FROM workspace.electiva.ordenes_mitad_anio
ORDER BY user_name, salesorderid;
```

```python
%sql
SELECT
    categoryname,
    `descuento_mitad_año`,
    COUNT(*) AS filas
FROM workspace.electiva.ordenes_mitad_anio
GROUP BY categoryname, `descuento_mitad_año`;
```

**PUNTO 4**

Su empresa quiere desplegar campañas publicitarias y de segmentación de clientes para 
incrementar las ventas del año 2011  y para ello usted debe  crear un pequeño sistema de 
clasificación de clientes que cumpla los siguientes requerimientos: 

1. Los clientes que hayan hecho 3 órdenes en el año son considerados clientes Oro. 

2. Los clientes que hayan hecho 2 órdenes en el año son considerados clientes Plata. 

3. Los clientes que hayan hecho menos de 2 órdenes al año son clientes Ocasional 

La cantidad de órdenes  que un cliente hizo en el año 2011  se obtiene contando la cantidad 
de salesorderid distintos, cada salesorderid es el identificador de una orden hecha por un 
cliente, queremos identificar la cantidad de salesorderid distintos para cada cliente, para así 
poder clasificarlos  en alguna de las categorías. 
Crear una vista materializada en el esquema, llamada customer_class que contenga el 
user_name  de  todos los clientes del año 2011 , una columna user_level donde indique si el 
cliente es “Oro”, “Plata” o “Ocasional”, además debe contener para cada cliente su 
customerid, su customer fullname y la cantidad de órdenes que realizó en el año 

```python
%sql
WITH conteo AS (
    SELECT
        c.CustomerID AS customerid,
        c.FullName AS customerfullname,
        u.user_name AS user_name,
        COUNT(DISTINCT o.SalesOrderID) AS amount_of_orders
    FROM workspace.electiva.orders AS o
    JOIN workspace.electiva.customers AS c ON o.CustomerID = c.CustomerID
    JOIN workspace.electiva.page_user AS u ON o.CustomerID = u.customerid
    WHERE o.OrderDate BETWEEN '2011-01-01' AND '2011-12-31'
    GROUP BY c.CustomerID, c.FullName, u.user_name
)
SELECT
    customerid,
    customerfullname,
    user_name,
    amount_of_orders,
    CASE
        WHEN amount_of_orders >= 3 THEN 'Oro'
        WHEN amount_of_orders = 2 THEN 'Plata'
        ELSE 'Ocasional'
    END AS user_level
FROM conteo;
```

```python
%sql
CREATE OR REPLACE MATERIALIZED VIEW workspace.electiva.customer_class AS
WITH conteo AS (
    SELECT
        c.CustomerID AS customerid,
        c.FullName AS customerfullname,
        u.user_name AS user_name,
        COUNT(DISTINCT o.SalesOrderID) AS amount_of_orders
    FROM workspace.electiva.orders AS o
    JOIN workspace.electiva.customers AS c ON o.CustomerID = c.CustomerID
    JOIN workspace.electiva.page_user AS u ON o.CustomerID = u.customerid
    WHERE o.OrderDate BETWEEN '2011-01-01' AND '2011-12-31'
    GROUP BY c.CustomerID, c.FullName, u.user_name
)
SELECT
    customerid,
    customerfullname,
    user_name,
    amount_of_orders,
    CASE
        WHEN amount_of_orders >= 3 THEN 'Oro'
        WHEN amount_of_orders = 2 THEN 'Plata'
        ELSE 'Ocasional'
    END AS user_level
FROM conteo;
```

```python
%sql
SELECT * FROM workspace.electiva.customer_class;
```

```python
%sql
SELECT user_level, COUNT(*) AS clientes
FROM workspace.electiva.customer_class
GROUP BY user_level;
```
